```python
import pandas as pd
import sqlite3

csv_file = 'electronics_sales_2024.csv'
df = pd.read_csv(csv_file)

db = 'sales_database.db'
conn = sqlite3.connect(db)

df.to_sql('sales_data', conn, if_exists='replace', index=False)

conn.close()

print(f"Successfully converted {csv_file} into {db}!")
```
